```python

import numpy as np

class BatchGradientDescent:

    def __init__(self, learning_rate=0.01, epochs=1000):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):
        X_train = np.asarray(X_train, dtype=float)
        y_train = np.asarray(y_train, dtype=float).reshape(-1)

        n_samples, n_features = X_train.shape

        self.intercept_ = 0.0
        self.coef_ = np.zeros(n_features)

        for _ in range(self.epochs):
            # Predict for all training examples
            y_hat = np.dot(X_train, self.coef_) + self.intercept_

            # Calculate gradients using the whole dataset
            intercept_der = -2 * np.mean(y_train - y_hat)
            coef_der = (-2 / n_samples) * np.dot((y_train - y_hat), X_train)

            # Update parameters
            self.intercept_ -= self.lr * intercept_der
            self.coef_ -= self.lr * coef_der

        return self

    def predict(self, X_test):
        X_test = np.asarray(X_test, dtype=float)
        return np.dot(X_test, self.coef_) + self.intercept_


# Example usage
X = np.array([[1], [2], [3], [4], [5]], dtype=float)
y = np.array([20, 40, 60, 80, 100], dtype=float)

model = BatchGradientDescent(learning_rate=0.01, epochs=1000)
model.fit(X, y)

print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
print("Predictions:", model.predict([[6], [7]]))
```
